```python
import pandas as pd
from pathlib import Path
file_path=Path('netflix_titles.csv')
print(f"loading files: {file_path}")
```

```python
df=pd.read_csv(file_path)

print(f'loaded {len(df)} rows and {len(df.columns)} columns')
df1=df.copy()
df2=df.copy()
```

```python
print(f"columns: {df.columns}")
df.info()
```

```python
df.isnull().sum().sort_values(ascending=False)  
```

```python
df1.dropna(subset=['director','country','cast','date_added','rating','duration'],inplace=True) 
df1.shape
```

```python
df1.isnull().sum().sort_values(ascending=False)
```

```python
print(df2.head(10))
```

```python

df2["director"] = df2["director"].fillna("Unknown")
df2["country"] = df2["country"].fillna("Unknown")
df2["cast"] = df2["cast"].fillna("Unknown")


df2["rating"] = df2["rating"].fillna(df2["rating"].mode()[0])


df2 = df2.dropna(subset=["date_added", "duration"])
```

```python
df2.isnull().sum().sort_values(ascending=False)
```

```python
print(df2.shape)
```

```python
df2.duplicated(subset=['show_id']).sum()
df2.apply(lambda col: col.duplicated().sum())
```

```python
df2.to_csv('cleaned_netflix_titles.csv', index=False)
```
